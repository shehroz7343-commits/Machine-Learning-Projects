```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
loan_dataset = pd.read_csv('loan_dataset.csv')
```

```python
loan_dataset = loan_dataset.dropna()
```

```python
loan_dataset.isnull().sum()
```

```python
loan_dataset.info()
```

```python
loan_dataset.head()
```

```python
loan_dataset['Loan_Status'].value_counts()
```

```python
loan_dataset.replace({
    'Loan_Status': {'N':0, 'Y':1}
}, inplace=True)
```

```python
loan_dataset.head()
```

```python
loan_dataset['Dependents'].value_counts()
```

```python
loan_dataset.replace({
    'Dependents': {'3+': 4}
}, inplace=True)
```

```python
loan_dataset['Dependents'].value_counts()
```

```python
loan_dataset.head()
```

```python
loan_dataset['Gender'].value_counts()
```

```python
loan_dataset['Married'].value_counts()
```

```python
loan_dataset['Education'].value_counts()
```

```python
loan_dataset['Self_Employed'].value_counts()
```

```python
loan_dataset['Property_Area'].value_counts()
```

```python
loan_dataset.replace({
    'Gender': {'Male': 0, 'Female': 1},
      'Married': {'Yes': 1, 'No': 0},
    'Education': {'Graduate': 0, 'Not Graduate': 1},
    'Self_Employed': {'Yes': 1, 'No': 0},
    'Property_Area': {'Semiurban': 0, 'Urban': 1, 'Rural': 2}
}, inplace=True)
```

```python
loan_dataset.head()
```

```python
X = loan_dataset.drop(columns=['Loan_ID', 'Loan_Status'], axis=1)
y = loan_dataset['Loan_Status']
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = svm.SVC(kernel='linear')
```

```python
model.fit(X_train, y_train)
```

```python
training_data_prediction = model.predict(X_train)
```

```python
training_data_accuracy = accuracy_score(training_data_prediction, y_train)
```

```python
print('Accuracy on training data: ', training_data_accuracy)
```

```python
testing_data_prediction = model.predict(X_test)
```

```python
testing_data_accuracy = accuracy_score(testing_data_prediction, y_test)
```

```python
print('Accuracy on testing data: ', testing_data_accuracy)
```
