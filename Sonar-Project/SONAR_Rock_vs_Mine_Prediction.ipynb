```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
sonar_dataset = pd.read_csv('Copy of sonar data.csv', header=None)
```

```python
X = sonar_dataset.drop(columns=60, axis=1)
y = sonar_dataset[60]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
training_data_prediction = model.predict(X_train)
```

```python
training_data_accuracy = accuracy_score(training_data_prediction, y_train)
```

```python
print('Accuracy on training data : ', training_data_accuracy)
```

```python
testing_data_prediction = model.predict(X_test)
```

```python
testing_data_accuracy = accuracy_score(testing_data_prediction, y_test)
```

```python
print('Accuracy on testing data : ', testing_data_accuracy)
```
